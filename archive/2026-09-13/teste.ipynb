```python
import numpy as np


z = 2.0
dist = 10.0

rbf = np.exp(-(z * dist) ** 2)
rbf
```

```python
def _mother_function(r: np.ndarray, r0: float) -> np.ndarray:
        return np.exp(-0.5 * (r / r0) ** 2)

def compute_weights(dists: np.ndarray, z) -> np.ndarray:
    r0 = z
    weights = _mother_function(dists, r0=r0)

    phi_sum = np.sum(weights)
    if phi_sum == 0:
        return np.ones(len(dists)) / len(dists)  # fallback uniforme

    weights /= phi_sum  # particao de unidade
    return weights

dists = np.array([0.0, 10.0, 20.0, 30.0])
print(compute_weights(dists=dists,z= 1.0))
print(compute_weights(dists=dists,z= 2.0))
print(compute_weights(dists=dists,z= 10.0))
```
